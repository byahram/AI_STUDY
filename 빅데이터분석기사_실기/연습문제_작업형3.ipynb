# 1. (가설검정) 단일 표본 검정

한 커피 제조회사에서는 새로 출시한 커피의 카페인 함량이 평균 95mg 미만이라고 주장하였다. 그 주장이 사실인가를 알아보기 위하여 25개의 커피 샘플을 무작위(랜덤)로 추출하였다. 커피 제조회사의 주장이 타당한지를 유의수준 5%에서 검정하시오.

```
귀무가설(H0): μ ≥ 95mg
대립가설(H1): μ < 95mg
```

````python
import pandas as pd
df = pd.DataFrame({
    'Caffeine(mg)': [
        94.2, 93.7, 95.5, 93.9, 94.0, 95.2, 94.7, 93.5, 92.8, 94.4,
        93.8, 94.6, 93.3, 95.1, 94.3, 94.9, 93.9, 94.8, 95.0, 94.2,
        93.7, 94.4, 95.1, 94.0, 93.6
    ]
})

# [1] 표본 평균
print(df['Caffeine(mg)'].mean())

# [2] Shapiro-Wilk의 p-value
from scipy import stats
print(stats.shapiro(df['Caffeine(mg)']))

# [3] t-검정의 검정통계량
statistic, pvalue = stats.ttest_1samp(df['Caffeine(mg)'], 95, alternative='less')
print(statistic)

# [4] t-검정의p-value
# print(pvalue)
print("{:.10f}".format(pvalue))

# [5] 귀무가설을 기준 채택/기각
print("귀무가설 기각" if pvalue < 0.05 else "귀무가설 채택")
````

# 2. (가설검정) 독립 표본 검정

스마트폰 배터리의 충전 시간을 줄이기 위한 새로운 충전기를 개발하였다. 개발된 충전기의 효과를 검정하기 위해 스마트폰 사용자 집단을 두 그룹으로 나누어 한 그룹에는 새로운 충전기를 제공하고, 나머지 그룹에는 기존의 충전기를 제공하였다. 그 후, 두 그룹의 평균 충전 완료 시간을 비교하였다. 두 그룹 모두 충전 시간은 정규분포를 따르며 분산은 같다고 가정한다. 새로운 충전기가 더 빠르다고 할 수 있는지를 유의수준 5%하에서 검정하시오. 여기서 두 집단은 모두 정규분포를 따르고 분산은 같다고 가정한다. (μ1: 새로운 충전기 집단의 모평균, μ2: 기존 충전기 집단의 모평균)

```
귀무가설(H0): μ1 ≥ μ2
대립가설(H1): μ1 < μ2
```

````python
# 데이터
import pandas as pd
df = pd.DataFrame({
    '충전기': ['New'] * 10 + ['Old'] * 10,
    '충전시간': [
        1.5, 1.6, 1.4, 1.7, 1.5, 1.6, 1.7, 1.4, 1.6, 1.5,
        1.7, 1.8, 1.7, 1.9, 1.8, 1.7, 1.8, 1.9, 1.7, 1.6
    ]
})
# print(df.head())


new_cond = (df['충전기'] == 'New')
old_cond = (df['충전기'] == 'Old')

from scipy import stats
t_stats, pvalue = stats.ttest_ind(df[new_cond]['충전시간'], df[old_cond]['충전시간'], alternative='less', equal_var=True)

# [1] 가설을 검정하기 위한 검정 통계량
print(t_stats)

# [2] 통계량에 대한 p-value
print(pvalue)

# [3] 귀무가설을 기준 채택/기각
print("귀무가설 기각" if pvalue < 0.05 else "귀무가설 채택")
````

# 3. (가설검정) 대응 표본 검정

딴짓 연구소에서는 새로 개발한 교육과정에 참여 하면 학습 시간이 단축한다고 주장한다. 10명의 학생들을 대상으로 기존의 교육 방법과 새로운 교육 방법을 적용한 결과에 대하여 유의수준 5%에서 새로운 교육 방법이 교육 시간을 단축시켰는지 검정하시오. (단, 모집단은 정규분포를 가정한다.)

```
μd = (새로운 교수 방법 – 기존 교수 방법)의 평균
귀무가설(H0): μd = 0
대립가설(H1): μd < 0
```

````python
# 데이터
import pandas as pd
df = pd.DataFrame({
    'User': list(range(1, 11)),
    '기존방법': [60.4, 60.7, 60.5, 60.3, 60.8, 60.6, 60.2, 60.5, 60.7, 60.4],
    '새로운방법': [59.8, 60.2, 60.1, 59.9, 59.7, 58.4, 57.0, 60.3, 59.6, 59.8]
})
# print(df.head(2))


# [1] μd의 표본평균
df['diff'] = df['새로운방법'] - df['기존방법']
print(df['diff'].mean())

# [2] 가설을 검정하기 위한 검정 통계량
from scipy import stats
statistics, pvalue = stats.ttest_rel(df['새로운방법'], df['기존방법'], alternative='less')
print(statistics)

# [3] 통계량에 대한 p-value
print(pvalue)

# [4] 귀무가설을 기준 채택/기각
print("귀무가설 기각" if pvalue < 0.05 else "귀무가설 채택")
````

# 4. (분산분석) 일원 분산 분석

4개의 중학교에서 다른 교육 방법을 사용하여 학생들에게 수학을 가르쳤습니다. 각 중학교의 학생들이 받은 교육 후의 수학 성적은 다음과 같습니다. 이 4개의 교육 방법에 따른 성적 차이가 통계적으로 유의미한지 5%의 유의 수준에서 검정하십시오.

```
귀무가설(H0): 모든 중학교의 학생들의 수학 성적 평균은 동일하다.
대립가설(H1): 적어도 두 개의 중학교의 학생들의 수학 성적 평균은 다르다. (4개 교육 방법 중 최소한 하나의 방법이 다른 방법들과 성적 차이가 있다.)
```

````python
import pandas as pd
# df = pd.read_csv("math.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part3/ch6/math.csv")
# print(df.head())
# print(df['groups'].value_counts())

from scipy import stats

# [1] Shapiro-Wilk 검정을 통한 정규성: p-value
condA = (df['groups'] == "group_A")
print(stats.shapiro(df[condA]['scores']))

condB = (df['groups'] == "group_B")
print(stats.shapiro(df[condB]['scores']))

condC = (df['groups'] == "group_C")
print(stats.shapiro(df[condC]['scores']))

condD = (df['groups'] == "group_D")
print(stats.shapiro(df[condD]['scores']))

# [2] 등분산성 Levene 검정: p-value
print(stats.levene(df[condA]['scores'], df[condB]['scores'], df[condC]['scores'], df[condD]['scores']))


# 일원분산분석 모델 학습
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

model = ols('scores ~ C(groups)', df).fit()
result = anova_lm(model)
print(result)

# [4] 그룹 변수의 자유도: 3.0
print(result.loc['C(groups)', 'df'])

# [5] 잔차의 자유도: 36.0
print(result.loc['Residual', 'df'])

# [6] 성적의 제곱합: 411.8
print(result.loc['C(groups)', 'sum_sq'])

# [7] 성적의 평균 제곱: 137.27
print(result.loc['C(groups)', 'mean_sq'])

# [8] F-통계량의 값: 34.17
# [9] 성적에 대한 p-value: 1.240642e-10
print(result.loc['C(groups)', 'F'])

# [3] 귀무가설 기각 기준 채택/기각
print("귀무가설 기각" if 1.240642e-10 < 0.05 else "귀무가설 채택")
````

# 5. (분산분석) 이원 분산 분석

H 토마토 연구소에서는 토마토의 수확량을 향상시키기 위해 3가지 다른 비료 유형과 4가지 다른 물 주기를 실험하고자 한다. 연구소에서는 12개의 토마토 식물 무작위로 선택하여 각 조합에 대해 반복적으로 실험을 수행하였다.

- 비료 유형에 따른 효과

    ```
    귀무가설(H0): 모든 비료 유형의 토마토 수확량 평균은 동일하다.
    대립가설(H1): 적어도 두 개 이상의 비료 유형에서의 토마토 수확량 평균이 다르다.
    ```

- 물 주기에 따른 효과

    ```
    귀무가설(H0): 모든 물 주기의 토마토 수확량 평균은 동일하다.
    대립가설(H1): 적어도 두 개 이상의 물 주기에서의 토마토 수확량 평균이 다르다.
    ```

- 비료 유형과 물 주기간의 상호작용 효과

    ```
    귀무가설(H0): 비료 유형과 물 주기 간의 상호작용은 토마토 수확량에 영향을 미치지 않는다.
    대립가설(H1): 비료 유형과 물 주기 간의 상호작용은 토마토 수확량에 영향을 미친다.
    ```

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p3/tomato2.csv")
# print(df.head())

from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

model = ols('수확량 ~ C(비료유형) * C(물주기)', df).fit()
result = anova_lm(model)
print(result)


# [1] 비료 유형에 따른 토마토의 수확량 평균에 차이가 있는지 검정통계량 : 3.184685
# [2] 위의 통계량에 대한 p-valu : 0.059334
# [3] 귀무가설을 기준 : 채택
# [4] 물 주기에 따른 토마토의 수확량 평균에 차이가 있는가 검정통계량 : 3.661490
# [5]. 위의 통계량에 대한 p-value : 0.026460
# [6] 귀무가설을 기준 : 기각
# [7] 비료 유형과 물 주기 간의 상호작용은 토마토 수확량에 영향이 있는가 검정 통계량 : 0.863491
# [8] 위의 통계량에 대한 p-value : 0.535426
# [9] 귀무가설을 기준 : 채택
````

# 6. (카이제곱) 적합도 검정

한 도시의 운전자 1,000명을 대상으로 교통사고 경험 수를 조사하였다. 1회: 550명, 2회: 250명, 3회 100명, 4회 70명, 5회 이상: 30명이다. 전국적으로 조사된 데이터에 따르면, 운전자들의 교통사고 경험 수 분포는 다음과 같다. 1회: 60%, 2회: 25%, 3회 8%, 4회 5%, 5회 이상: 2%이다. 이 도시 운전자들의 교통사고 경험 수 분포가 전국적인 경향을 따르는지 검정하시오. (유의수준 0.05)

```
귀무가설(H0): 이 도시의 교통사고 경험 수 분포는 전국적인 경향을 따른다.
대립가설(H1): 이 도시의 교통사고 경험 수 분포는 전국적인 경향을 따르지 않는다.
```

````python
import pandas as pd

# [1] 5회 이상의 교통사고 경험자 비율을 0과1 사이로
print(30 / 1000)

# [2] 교통사고 경험 수 분포가 전국적인 경향을 따르는지를 검정하기 위한 검정 통계량
from scipy.stats import chisquare
observed = [550, 250, 100, 70, 30]
expected = [1000*0.6, 1000*0.25, 1000*0.08, 1000*0.05, 1000*0.02]
statistic, pvalue = chisquare(observed, expected)
print(statistic)

# [3] 위의 통계량에 대한 p-value
print(pvalue)

# [4] 귀무가설 기준 채택/기각
print("귀무가설 기각" if pvalue < 0.05 else "귀무가설 채택")
````

# 8. 회귀

```
귀무가설(H0): 광고비의 회귀 계수는 0이다. (광고비가 배달 주문량에 영향을 주지 않는다.)
대립가설(H1): 광고비의 회귀 계수는 0이 아니다. (광고비가 배달 주문량에 영향을 준다.)
```

````python
# 데이터
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p3/order.csv")
# print(df.head())

from statsmodels.formula.api import ols
model = ols('주문량 ~ 할인율 + 온도 + 광고비', data=df).fit()

# [1] 할인율와 온도의 상관계수
print("[1] 상관계수:", round(df['할인율'].corr(df['온도']), 2))

# [2] 모델의 결정계수
print("[2] 결정계수(R-squared):", round(model.rsquared, 2))

# [3] 각 변수의 회귀 계수(기울기)
print("[3] 회귀계수:", round(model.params, 4))

# [4] 모델의 절편 (Intercept)
print("[4] 절편:", round(model.params['Intercept'], 4))

# [5] 온도의 회귀 계수가 통계적으로 유의한지 검정 (p-value)
print("[5] pvalue:", round(model.pvalues['온도'], 4))

# [6] 데이터가 할인율이 10%, 온도가 20°C, 광고비가 500일 때 배달 주문량 예측
new_data = pd.DataFrame({"할인율": [10], "온도": [20], "광고비": [500]})
result = model.predict(new_data)
print("[6] 예측 판매량:", int(result[0]))

# [7] 잔차 제곱합
df['잔차'] = df['주문량'] - model.predict(df)
print("[7] 잔차 제곱합:", round(sum(df['잔차']**2), 2))

# [8] MSE
MSE = (df['잔차'] ** 2).mean()
print('[8] MSE:', round(MSE, 4))

# [9] 온도의 회귀 계수에 대한 90% 신뢰구간
print("[9] 신뢰구간:\n", model.conf_int(alpha=0.1))

# [10] 데이터가 할인율이15%, 온도가 25°C, 광고비가 300일 때 90%의 신뢰구간과 예측구간
new_data = pd.DataFrame({"할인율": [15], "온도": [25], "광고비": [300]})
pred = model.get_prediction(new_data)
result = pred.summary_frame(alpha=0.1)
print("[10] 예측값의 신뢰구간과 예측구간:\n", result)

# [11] 독립변수 할인율과 온도를 고정한 상태에서, 광고비가 배달 주문량에 영향을 주는지 가설을 검정
print("[11] 귀무가설 기각" if model.pvalues['광고비'] < 0.05 else "[11] 귀무가설 채택")
````

# 9. 로지스틱 회귀

주어진 데이터를 사용하여 데이터를 앞 50%의 a와 나머지 50%의 b로 나누십시오. a 데이터로 로지스틱 회귀 모델을 적합하고, 다음 문제에 답하시오.

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p3/customer_travel.csv")
# print(df.head())


# 데이터 분할
midpoint = len(df) // 2 #몫(정수)
a = df.iloc[:midpoint]
b = df.iloc[midpoint:]


# 데이터 확인
# print(a.shape, b.shape)


# [1] 유의하지 않은 독립변수의 개수
from statsmodels.formula.api import logit
formula = "target ~ age + service + social + booked"
model = logit(formula, data=a).fit()
# print(model.summary())
print("[1]", sum(model.pvalues[1:] >= 0.05)) # 상수항(Intercept) 제외


# [2] p-value가 0.05보다 작은 유의한 변수만 사용하여 가장 큰 p-value를 가진 변수의 이름
formula = 'target ~ age + booked'
model = logit(formula, data=a).fit()
# print(model.summary())
print("[2]", model.pvalues[1:].idxmax()) # 상수항(Intercept) 제외


# [3] 수정된 모델에서 독립변수 중 절댓값이 가장 큰 회귀계수를 가진 변수의 이름
print("[3]", model.params[1:].abs().idxmax())


# [4] 로그 우도
print("[4]", model.llf)


# [5] 잔차이탈도
print("[5]", -2 * model.llf)


# [6] 'booked' 변수가 3 증가할 때 오즈비
import numpy as np
print("[6]", np.exp(model.params['booked'] * 3))


# [7] p-value가 0.05보다 작은 회귀계수의 총합
print("[7]", model.params[model.pvalues < 0.05].sum()) # 상수항(절편)도 포함


# [8] 정확도
pred = model.predict(b)
pred = (pred > 0.5).astype(int)
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(b['target'], pred)
print("[8]", accuracy)



# [9] 오류율
error_rate = 1 - accuracy
error_rate
print("[9]", error_rate)
````
