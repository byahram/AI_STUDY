### Section16. 이상치, IQR

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p1/type1_data1.csv")


# [1] 'views' 컬럼의 1사분위수, 3사분위수 그리고 IQR
q1 = df['views'].quantile(0.25)
q3 = df['views'].quantile(0.75)
iqr = q3 - q1
# print(q1, q3, iqr)

# [2] 이상치 조건에 맞는 데이터
# (이상치는 1사분위수 – (IQR * 1.5)보다 작은 값과 3사분위수 + (IQR * 1.5)보다 큰 값)
line1 = q1 - (iqr * 1.5)
line2 = q3 + (iqr * 1.5)

cond1 = df['views'] < line1
cond2 = df['views'] > line2

df = df[cond1 | cond2]

# [3] 이상치 데이터의 'views' 컬럼 합을 정수
print(int(df['views'].sum()))
````

### Section17. 이상치, 소수점 있는 데이터 찾기, 표준편차

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p1/type1_data1.csv")


# [1] ‘views’ 컬럼의 표준편차
r1 = df['views'].std()
print(r1)
 
# [2] ‘age’ 컬럼의 이상치(소수점 나이, 음수 나이, 0살)를 제거하고, ‘views’ 컬럼의 표준편차
cond1 = (df['age'] > 0)
cond2 = (df['age'] == round(df['age'], 0)) # 소수점 이하가 0인 값(정수값)만 선택
df = df[cond1 & cond2]
r2 = df['views'].std()
print(r2)

# [3] 이상치 제거 전후의 ‘views’ 컬럼의 표준편차를 더하여, 반올림 후 소수 둘째 자리까지
print(round(r1 + r2, 2))
````

### Section22. 시간 간의 차이 계산(분), 그룹핑

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p1/delivery_time.csv")
# print(df.head())
# print(df.info())


# [1] 주문 시간과 실제 도착 시간의 차이를 분 단위로 계산
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])
df['diff'] = (df['실제도착시간'] - df['주문시간']).dt.total_seconds() / 60
# print(df['diff'])
 
# [2] 앱 종류별로 평균 도착 시간(분)을 계산
df = df.groupby('앱종류')['diff'].mean()
print(df.head())

# [3] 평균적으로 가장 빠른 앱 종류를 찾고, 해당 앱의 평균 도착 시간을 분으로 반올림하여 정수
print(round(df.min()))
````

### Section23. 시간 간의 차이 계산(분), 비율

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p1/delivery_time.csv")
# print(df.head())
# print(df.info())


# [1] 각 결제 종류별로 실제 도착 시간이 예상 도착 시간보다 늦은 주문의 비율
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'])
df['지연여부'] = (df['실제도착시간'] - df['예상도착시간']).dt.total_seconds() / 60
df['지연여부'] = df['지연여부'] > 0
result = df.groupby('결제종류')['지연여부'].mean()
# print(result)

# [2] 비율 중 가장 큰 값을 반올림하여 소수 둘째 자리까지 
print(round(result.max(), 2))
````

### Section24. 그룹핑, 값 찾기, 필터링

### Section27. 시간 범위, 속도(km/h)

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p1/delivery_time.csv")


# [1] 점심시간(10시부터 13시 전까지)에 주문된 배달 데이터
df['주문시간'] = pd.to_datetime(df['주문시간'])
cond = (df['주문시간'].dt.hour >= 10) & (df['주문시간'].dt.hour < 13)
# print(sum(cond))
df = df[cond]
# print(df.head())

# [3] 배달시간 = 실제도착시간 - 주문시간
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])
df['배달시간'] = df['실제도착시간'] - df['주문시간']
df['배달시간'] = df['배달시간'].dt.total_seconds() / 60 / 60
# print(df['배달시간'].info())

# [4] 속도(km/h) = 거리(km) / 시간(h)
df['속도'] = df['거리'] / df['배달시간']

# [2] 점심시간 주문 건 중 과속(평균 속도가 50km/h 이상)하는 주문 수를 정수로
print(sum(df['속도'] >= 50))
````

### Section29. 함수, 월별 집계

```
[배달료 기준표]
- 5km 미만: 2000원
- 5km 이상 ~ 10km 미만: 4000원
- 10km 이상 ~ 15km 미만: 6000원
- 15km 이상 ~ 20km 미만: 8000원
```

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p1/delivery_time.csv")
# print(df.head())


# [1] 배달료 계산 기준표에 따라 각 주문에 대한 배달료
def delivery_fee(distance):
    if distance < 5:
        return 2000
    elif distance < 10:
        return 4000
    elif distance < 15:
        return 6000
    elif distance < 20:
        return 8000

df['delivery_fee'] = df['거리'].apply(delivery_fee)
# print(df['delivery_fee'])

# [2] 월별로 배달료의 총합
df['주문시간'] = pd.to_datetime(df['주문시간'])
m = df['주문시간'].dt.to_period('M')
# print(m)
monthly_fee = df.groupby(m)['delivery_fee'].sum()
# print(monthly_fee)

# [3] 배달료가 가장 많이 발생한 월을 찾고, 그 월의 총 배달료를 정수로
monthly_fee = monthly_fee.sort_values(ascending=False)
print(monthly_fee[monthly_fee.idxmax()])
````

### Section36. 조건별 변환 (transform)

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/refs/heads/main/part1/ch3/sales.csv")
# print(df.head())


# [1] 결측된 판매금액은 해당 지역의 평균 판매금액으로 결측값을 대체
m = df.groupby('지역코드')['판매금액'].transform('mean')
# print(m)
df['판매금액'] = df['판매금액'].fillna(m)
# print(df['판매금액'])

# [2] 각 거래마다 “판매금액”과 해당 지역의 평균 판매금액의 "차이"를 절대값으로
df['지역평균'] = df.groupby('지역코드')['판매금액'].transform('mean')
df['차이'] = abs(df['판매금액'] - df['지역평균'])
# print(df['차이'])

# [3] 각 지역에서 차이값의 평균을 구한 후, 이 값이 가장 큰 지역의 지역코드
print(df.groupby('지역코드')['차이'].mean().idxmax())
````

### Section37. 재구조화 (unstack), 맵핑

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/refs/heads/main/part1/ch3/store_sales.csv")
# print(df.head())


# [1] 각 행에서 판매수량과 단가를 이용한 매출액 계산
df['매출액'] = df['판매수량'] * df['단가']
# print(df['매출액'])

# [2] 요일을 평일과 주말로 구분하고, 매장별 평일과 주말 매출액 합계
day_mapping = {'월': '평일', '화': '평일', '수': '평일', '목': '평일', '금': '평일', '토': '주말', '일': '주말'}
df['구분'] = df['요일'].map(day_mapping)
# print(df.head())
store_sales = df.groupby(['매장코드', '구분'])['매출액'].sum().unstack()
# print(store_sales)

# [3] 매장별 평일과 주말 매출액 차이를 절대값으로 + 모든 매장 중 가장 큰 절대값 차이
store_sales['차이'] = abs(store_sales['평일'] - store_sales['주말'])
print(store_sales['차이'].max())
````

### Section38. 피벗테이블

````python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/refs/heads/main/part1/ch3/region_sales.csv")


# [1] 각 Region과 Channel 조합별로 제품(Product) A, B의 총 판매액
pivot_table = pd.pivot_table(df, index=['Region', 'Channel'], columns='Product', values='Sales', aggfunc='sum')
# print(pivot_table)

# [2] 제품 A의 매출 비율(A비율)을 구하세요. 
#     A비율 = (제품 A 판매액) ÷ (제품 A 판매액 + 제품 B 판매액)
pivot_table['총매출'] = pivot_table['A'] + pivot_table['B']
pivot_table['A비율'] = pivot_table['A'] / pivot_table['총매출']
# print(pivot_table['총매출'], pivot_table['A비율'])

# [3] A비율 중 최댓값을 찾아 반올림하여 소수 둘째 자리까지
print(round(pivot_table['A비율'].max(), 2))
````
